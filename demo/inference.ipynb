```python
from src.preprocessing.preprocessing import processor
from src.models.lstm import LSTMModel
from src.utils.helpers import create_sequences
from src.postprocessing.inference import FloodClassifier
```

```python
classifier = FloodClassifier()
pcr = processor.load(path="demo", name="processor")
lstm = LSTMModel.load_model(name="lstm_model.keras", override_path="demo")
```

```python
inference_df = pcr.prep_inference(wandb_config={
    "file_path": "flood-dataset-top30",
    "file_name": "flood_model_top30",           # optional
    "start_date": "2023-12-01",                 # optional
    "end_date":   "2023-12-06",                 # optional
    "frequency":  "hourly",                     # optional
})
```

```python
X_data, _, train_site_ids = create_sequences(inference_df, None, 72)
preds = lstm.predict(X_data[:2])
unscaled_preds = pcr.unscale(preds, train_site_ids)
results = classifier.classify(unscaled_preds, train_site_ids)
results
```
