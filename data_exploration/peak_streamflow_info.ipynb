```python
from SRC.helper_functions.helpers import pull_duckdb
sites = ["06926510",
"06821250",
"06927000",
"06934000",
"06910450",
"06893500",
"06893000"]
df = pull_duckdb(file_name="wandb.flood_model", sites=sites, start_date="2010-01-01", end_date="2025-12-31")
```

```python
df.unique("site_id") #validate all columns are getting pulled correctly
```

```python
import polars as pl

numeric_cols = [c for c in df.columns if df[c].dtype in (pl.Float32, pl.Float64, pl.Int32, pl.Int64)]

for site in sites:
    site_df = df.filter(pl.col("site_id") == site)
    # min-max normalize numeric columns to [0, 1]
    normalized = site_df.select(
        [pl.col(c) for c in site_df.columns if c not in numeric_cols]
        + [
            ((pl.col(c) - pl.col(c).min()) / (pl.col(c).max() - pl.col(c).min())).alias(c)
            for c in numeric_cols
        ]
    )
    desc = normalized.describe()
    # keep only mean and std rows
    desc = desc.filter(pl.col("statistic").is_in(["mean", "std"]))
    # filter out columns with std < .1
    std_row = desc.filter(pl.col("statistic") == "std")
    keep_cols = [
        c for c in desc.columns
        if c == "statistic"
        or (std_row[c][0] is not None and float(std_row[c][0]) >= 0.1)
    ]
    desc = desc.select(keep_cols)
    # sort remaining columns by std (descending)
    col_order = sorted(
        [c for c in keep_cols if c != "statistic"],
        key=lambda c: float(std_row[c][0]),
        reverse=True,
    )
    print(f"\nSite: {site}")
    print(desc.select(["statistic"] + col_order))
```

```python
from sklearn.feature_selection import mutual_info_regression
import polars as pl

target_col = "streamflow_cfs_mean"
feature_cols = [c for c in numeric_cols if c != target_col]

for site in sites:
    site_df = df.filter(pl.col("site_id") == site).drop_nulls(subset=[target_col] + feature_cols)
    if site_df.height == 0:
        print(f"\nSite: {site} — no data after dropping nulls")
        continue

    X = site_df.select(feature_cols).to_numpy()
    y = site_df[target_col].to_numpy()

    mi_scores = mutual_info_regression(X, y, random_state=42)
    mi_df = (
        pl.DataFrame({"feature": feature_cols, "mutual_info": mi_scores})
        .sort("mutual_info", descending=True)
    )
    print(f"\nSite: {site} — Mutual Information with {target_col}")
    print(mi_df)
```
